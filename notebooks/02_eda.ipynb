Explores transactional distributions, seasonal sales velocity, regional distributions, and line-item basket sizes.

### 1. Load Processed Analytical Dataset

```
import sys
from pathlib import Path
import pandas as pd
import plotly.express as px

BASE_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(BASE_DIR))

from src.data_loader import load_processed_data
df = load_processed_data()
df.info()
```

### 2. Line Item Pricing & Revenue Distribution

```
fig_hist = px.histogram(df, x='net_revenue', nbins=50, title='Distribution of Line-Item Net Revenue', template='plotly_white')
fig_hist.show()
```

### 3. Order Volume by Category & Region

```
cat_reg = df.groupby(['region', 'category'])['net_revenue'].sum().reset_index()
fig_bar = px.bar(cat_reg, x='region', y='net_revenue', color='category', barmode='group',
                 title='Net Revenue by Region & Category', template='plotly_white')
fig_bar.show()
```
