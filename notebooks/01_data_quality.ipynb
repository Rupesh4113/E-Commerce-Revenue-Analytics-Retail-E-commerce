### 1. Ingestion of Raw Sources & Setup

```
import sys
from pathlib import Path
import pandas as pd
import json

# Setup project root path
BASE_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(BASE_DIR))

from src.data_loader import load_raw_orders, load_raw_customers
from src.data_cleaning import inspect_schema, clean_data

raw_orders = load_raw_orders()
raw_customers = load_raw_customers()
print(f"Loaded Raw Orders: {raw_orders.shape[0]:,} rows, {raw_orders.shape[1]} columns")
print(f"Loaded Raw Customers: {raw_customers.shape[0]:,} rows, {raw_customers.shape[1]} columns")
```

### 2. Schema Inspection & Missing Values Profile

```
schema_report = inspect_schema(raw_orders)
print("Total records:", schema_report['total_rows'])
print("Total columns:", schema_report['total_columns'])
missing_df = pd.DataFrame([
    {'column': k, 'null_count': v['null_count'], 'null_pct': v['null_percentage']}
    for k, v in schema_report['columns'].items() if v['null_count'] > 0
])
missing_df
```

### 3. Data Cleaning Pipeline Execution

```
cleaned_orders, dq_report = clean_data(raw_orders, raw_customers)
print("Initial rows:", dq_report['initial_rows'])
print("Cleaned rows:", dq_report['final_rows'])
print("Duplicates dropped:", dq_report['duplicates_removed'])
print("Invalid records dropped:", dq_report['invalid_records_removed'])
print("Date coverage:", dq_report['date_range'])
print("Unique entities:", dq_report['uniques'])
```
