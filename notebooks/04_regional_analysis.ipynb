### 3. Logistics Corridor Delivery Efficiency

```
corridor_summary = df.groupby(['region', 'logistics_hub']).agg(
    orders=('order_id', 'nunique'),
    avg_transit_days=('transit_days', 'mean'),
    return_rate=('return_flag', 'mean'),
    dispute_rate=('dispute_flag', 'mean')
).reset_index()
display(corridor_summary)
```
